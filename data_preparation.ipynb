```python
# Import required libraries
import os
import pandas as pd

# Define the path to the original IMDB dataset
DATASET_PATH = '/content/drive/MyDrive/Sentiment_Analysis/dataset'

# Define where the combined CSV will be saved
OUTPUT_PATH = '/content/drive/MyDrive/NLP/Sentiment_Analysis/combined_dataset/imdb_combined.csv'


# Function to load all reviews from one train/test folder
def load_imdb_folder(folder_path):

    # Create a list to store all reviews
    reviews = []

    # Create a list to store the corresponding sentiment labels
    sentiments = []

    # Loop through positive and negative folders
    for sentiment in ['pos', 'neg']:

        # Create the path to the sentiment folder
        sentiment_path = os.path.join(folder_path, sentiment)

        # Get all TXT filenames
        filenames = [
            filename
            for filename in os.listdir(sentiment_path)
            if filename.endswith('.txt')
        ]

        # Read every review file
        for filename in filenames:

            # Create the complete file path
            file_path = os.path.join(sentiment_path, filename)

            # Read the review text
            with open(file_path, 'r', encoding='utf-8') as file:
                review = file.read()

            # Store the review
            reviews.append(review)

            # Store the sentiment
            sentiments.append(sentiment)

    # Return the reviews and labels
    return reviews, sentiments


# Create empty lists for the complete dataset
all_reviews = []
all_sentiments = []


# Load the original training reviews
train_reviews, train_sentiments = load_imdb_folder(
    os.path.join(DATASET_PATH, 'train')
)

# Add training data to the complete dataset
all_reviews.extend(train_reviews)
all_sentiments.extend(train_sentiments)


# Load the original testing reviews
test_reviews, test_sentiments = load_imdb_folder(
    os.path.join(DATASET_PATH, 'test')
)

# Add testing data to the complete dataset
all_reviews.extend(test_reviews)
all_sentiments.extend(test_sentiments)


# Create one DataFrame containing all 50,000 reviews
df = pd.DataFrame({
    'review': all_reviews,
    'sentiment': all_sentiments
})


# Save only the review and sentiment columns
df.to_csv(
    OUTPUT_PATH,
    index=False
)


# Display the results
print("Dataset created successfully!")
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
print("Saved to:", OUTPUT_PATH)
```
